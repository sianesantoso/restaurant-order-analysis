```python
# import libraries
import pandas as pd
from titlecase import titlecase
from sqlalchemy import create_engine
```

```python
# load dataset
data = pd.read_csv(r"../Dataset/Raw Data/menu_items.csv")
```

```python
# data overview
print("-" * 40)
print("DATA OVERVIEW")

# rows and columns 
print("-" * 40)
print(f"Total Baris : {data.shape[0]}")
print(f"Total Kolom : {data.shape[1]}")

# columns names
print("-" * 40)
print("Nama Kolom")
for index, column in enumerate(data.columns, start = 1) :
    print(f"{index}. {column}")

# data types
print("-" * 40)
display(pd.DataFrame({
    "Nama Kolom" : data.dtypes.index,
    "Tipe Data" : data.dtypes.values
}).style.hide(axis = "index"))

# priview dataset
data.head(3)
```

```python
# missing values
print("-" * 40)
print("MISSING VALUES")

# check na values per column
print("-" * 40)
display(pd.DataFrame({
    "Nama Kolom" : data.isna().sum().index,
    "Total NA" : data.isna().sum().values
}).style.hide(axis = "index"))
```

```python
# duplicate values 
print("-" * 40)
print("DUPLICATE VALUES")

print("-" * 40)
print(f"Total duplikat : {data.duplicated().sum()}")
```

```python
# text Standarization
print("-" * 40)
print("TEXT STANDARIZATION")

print("-" * 40)
string_columns = ["item_name", "category"]
for i in string_columns :
    # trim leading and trailing white space 
    data[i] = data[i].str.strip()
    # standardizing letter case
    data[i] = data[i].apply(titlecase)
```

```python
# data type validation
print("-" * 40)
print("DATA TYPE VALIDATION")
print("-" * 40)

# data type alr valid 
```

```python
# data validation
print("-" * 40)
print("DATA VALIDATION")
print("-" * 40)

# check anomalous values 
print(f"Total Data Sales < 0 : {len(data[data["price"] < 0])}")
```
