```python
# import libraries
import pandas as pd
from titlecase import titlecase
from sqlalchemy import create_engine
```

```python
# load dataset
data = pd.read_csv(r"../Dataset/Raw Data/order_details.csv")
```

```python
# data overview
print("-" * 40)
print("DATA OVERVIEW")

# rows and columns 
print("-" * 40)
print(f"Total Baris : {data.shape[0]}")
print(f"Total Kolom : {data.shape[1]}")

# columns names
print("-" * 40)
print("Nama Kolom")
for index, column in enumerate(data.columns, start = 1) :
    print(f"{index}. {column}")

# data types
print("-" * 40)
display(pd.DataFrame({
    "Nama Kolom" : data.dtypes.index,
    "Tipe Data" : data.dtypes.values
}).style.hide(axis = "index"))

# priview dataset
data.head(6)
```

```python
"""
Since the Order_Details doesn't contain records that match the corresponding menu names in the Menu File, 
these records cannot be linked through the intended key. 
Therefore, the unmatched records in OrderDetails are considered invalid or canceled transactions and are excluded from further analysis project.
"""

# missing values
print("-" * 40)
print("MISSING VALUES")

# check na values per column
print("-" * 40)
print(f"Before Checking NA")
display(pd.DataFrame({
    "Nama Kolom" : data.isna().sum().index,
    "Total NA" : data.isna().sum().values
}).style.hide(axis = "index"))

# drop rows with na item_id 
data.dropna(subset = ["item_id"], inplace = True)

# re-check na values per column
print("-" * 40)
print(f"After Checking NA")
display(pd.DataFrame({
    "Nama Kolom" : data.isna().sum().index,
    "Total NA" : data.isna().sum().values
}).style.hide(axis = "index"))
```

```python
# duplicate values 
print("-" * 40)
print("DUPLICATE VALUES")

print("-" * 40)
print(f"Total duplikat : {data.duplicated().sum()}")
```

```python
# text Standarization
print("-" * 40)
print("TEXT STANDARIZATION")
print("-" * 40)

# no string columns should be standardized
```

```python
# data type validation
print("-" * 40)
print("DATA TYPE VALIDATION")

# before standardized
print("-" * 40)
print("Data Type Before Standarized : ")
display(pd.DataFrame({
    "Nama Kolom" : data.dtypes.index,
    "Tipe Data" : data.dtypes.values
}).style.hide(axis = "index"))

# standardrize order_date dan order_time
data["order_date"] = pd.to_datetime(data['order_date'], format='%m/%d/%y').dt.date
data["order_time"] = pd.to_datetime(data["order_time"], format='%I:%M:%S %p').dt.time
data["item_id"] = data["item_id"].astype(int)

# re-check after standardized
print("-" * 40)
print("Data Type After Standarized : ")
display(pd.DataFrame({
    "Nama Kolom" : data.dtypes.index,
    "Tipe Data" : data.dtypes.values
}).style.hide(axis = "index"))
```

```python
# data validation
print("-" * 40)
print("DATA VALIDATION")
print("-" * 40)

# no anomalous values
```
